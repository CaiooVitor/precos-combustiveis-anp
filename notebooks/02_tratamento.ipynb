```python
import pandas as pd
```

```python
sem1_2025 = pd.read_csv("../dados/brutos/ca-2025-01.csv", sep=";", encoding="utf-8-sig", low_memory=False)
sem2_2025 = pd.read_csv("../dados/brutos/ca-2025-02.csv", sep=";", encoding="utf-8-sig",low_memory=False)
sem1_2026 = pd.read_csv("../dados/brutos/ca-2026-01.csv", sep=";", encoding="utf-8-sig",low_memory=False)
```

```python
df = pd.concat([sem1_2025, sem2_2025, sem1_2026], ignore_index=True)
df.shape
```

```python
df["Valor de Venda"] = df["Valor de Venda"].str.replace(",",".").astype(float)
df["Valor de Venda"].describe()
```

```python
df["Valor de Venda"].isnull().sum()
```

```python
df = df.dropna(subset=["Valor de Venda"])
df.shape
```

```python
df.duplicated().sum()
```

```python
df[["Municipio", "Estado - Sigla", "Produto", "Data da Coleta"]].isnull().sum()
```

```python
df = df.drop_duplicates()
```

```python
df["Data da Coleta"] = pd.to_datetime(df["Data da Coleta"], format="%d/%m/%Y")
df.info()
```

```python
df = df.drop(columns=["Valor de Compra"])
```

```python
print(df["Estado - Sigla"].unique())
print(df["Produto"].unique())
print(df["Unidade de Medida"].unique())
print(df["Bandeira"].unique())
```

```python
df["Unidade de Medida"] = df["Unidade de Medida"].replace("R$ / m3", "R$ / m³")
```

```python
df.to_csv("../dados/tratados/precos_combustiveis_tratados.csv", index=False, encoding="utf-8-sig")
```
