```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
df = pd.read_csv("../dados/tratados/precos_combustiveis_tratados.csv", encoding="utf-8-sig")
df.shape
```

```python
engine = create_engine("sqlite:///../dados/tratados/combustiveis.db")
df.to_sql("precos_combustiveis", engine, if_exists="replace", index=False)
```

```python
resultado = pd.read_sql("SELECT * FROM precos_combustiveis LIMIT 5", engine)
resultado
```
